##  Correlation Analysis

```python

sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.show()
```

## Correlation Analysis Insights

- Attendance has the strongest impact on exam performance (correlation: 0.58), making it the most critical factor.
- Study hours also significantly influence scores (0.45), highlighting the importance of consistent study habits.
- Previous scores and tutoring sessions show moderate influence, indicating that academic background and additional support play a role.
- Physical activity and sleep hours show minimal correlation, suggesting they may not directly affect exam performance in this dataset.

##  How the Model Works

The model predicts exam scores based on study hours, attendance, and previous scores.

- 📘 Study Hours: More study time increases the predicted score significantly.
- 🏫 Attendance: Regular attendance improves performance.
- 📊 Previous Scores: Past performance has a smaller impact.

👉 The model combines these factors to estimate the final exam score.

```python
from sklearn.linear_model import LinearRegression

X = df[["Hours_Studied", "Attendance", "Previous_Scores"]]
y = df["Exam_Score"]

model = LinearRegression()
model.fit(X, y)
model.coef_
```

```python
features = ["Hours_Studied", "Attendance", "Previous_Scores"]
importance = model.coef_

import matplotlib.pyplot as plt

plt.bar(features, importance)
plt.xlabel("Features")
plt.ylabel("Impact on Score")
plt.title("Feature Importance")
plt.show()
```

Study hours have the highest impact on exam score, followed by attendance, while previous scores have minimal influence.

###example


```python
sample = [[5, 80, 60]]
prediction = model.predict(sample)

print("Predicted Exam Score:", prediction[0])
```
